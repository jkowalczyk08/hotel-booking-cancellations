# Create Data Splits

This notebook splits the cleaned dataset into Training, Validation, and Test sets.
We use a **70% / 15% / 15%** split ratio, stratified by the target variable `is_canceled`.

Splits are saved using parquet.

```python
import sys
import os
import pandas as pd
from sklearn.model_selection import train_test_split

sys.path.append(os.path.abspath('..'))

from src.set_utils import load_hotel_data, save_splits
```

```python
DATA_PATH = '../data/hotel_bookings_cleaned.csv'
df = load_hotel_data(DATA_PATH)
```

```python
SEED = 42

X = df.drop('is_canceled', axis=1)
y = df['is_canceled']

X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.15, stratify=y, random_state=SEED
    )

val_size = 0.15 / 0.85 # 15% of total

X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=val_size, stratify=y_temp, random_state=SEED
    )

print(f"Train Share: {len(X_train)/len(df):.2%}")
print(f"Val Share:   {len(X_val)/len(df):.2%}")
print(f"Test Share:  {len(X_test)/len(df):.2%}")
```

```python
OUTPUT_DIR = '../data/'
save_splits(X_train, y_train, X_val, y_val, X_test, y_test, OUTPUT_DIR)
```
